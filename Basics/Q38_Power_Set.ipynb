```python
def allPossibleStrings(s):
    n = len(s)
    result = []
    for i in range(1 << n):   
        subset = ""
        for j in range(n):
            if i & (1 << j):
                subset += s[j]
        result.append(subset)   

    return result

s = "abc"
subsets = allPossibleStrings(s)
 
for subset in subsets:
    print(subset)
```
