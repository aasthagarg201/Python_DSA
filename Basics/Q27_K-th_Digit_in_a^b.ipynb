```python
def kthdigit(a, b, k):
    mod = 10 ** k
    res = 1
    base = a

    while b > 0:
        if b & 1:
            res = (res * base) % mod
        base = (base * base) % mod
        b >>= 1
    for i in range(1, k):
        res //= 10
    return res

a = 5
b = 2
k = 1
print(kthdigit(a, b, k))
```
